# PID Speed Control

short CARLA run for checking longitudinal PID controller. The car stays on a straight road and only throttle/brake are controlled.


## Setup



```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
candidates = [cwd, cwd / 'carla_controllers', cwd.parent]
PROJECT_ROOT = next(
    path for path in candidates
    if (path / 'configs' / 'pid_speed.yaml').is_file()
)
SRC_DIR = PROJECT_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

CONFIG_PATH = PROJECT_ROOT / 'configs' / 'pid_speed.yaml'
print(f'Controller root: {PROJECT_ROOT}')
print(f'Configuration: {CONFIG_PATH}')
```

## PID Check

YAML file stores the gains and speed profile. Positive controller output becomes throttle and negative output becomes brake.


```python
from common import load_yaml, validate_pid_config

config = load_yaml(CONFIG_PATH)
validate_pid_config(config)
config['control']
```

## Run

The run creates a timestamped folder with telemetry, the used config, JSON summary.


```python
from run_pid_speed import run_experiment

summary = run_experiment(CONFIG_PATH)
summary
```
